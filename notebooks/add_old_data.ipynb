# Adding data from previous month and months

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load the data
df = pd.read_csv('../data/processed/final_3d_df.csv')
```

```python

df = df.sort_values(['fips', 'year', 'month']).reset_index(drop=True)

lag_columns = ['Affordability Metric', 'county_perc_mo', 'mortgage_rate', 'income']

for col in lag_columns:
    df[f'{col}_prev_mo'] = df.groupby('fips')[col].shift(1)

df = df.dropna()
df.head()
```


## Add Rolling Averages (3-Month and 12-Month)


```python

rolling_columns = ['Affordability Metric', 'county_perc_mo', 'mortgage_rate', 'income']

# 3-month rolling average
for col in rolling_columns:
    df[f'{col}_rolling_3m'] = df.groupby('fips')[col].transform(lambda x: x.rolling(window=3, min_periods=1).mean())

# 12-month rolling average
for col in rolling_columns:
    df[f'{col}_rolling_12m'] = df.groupby('fips')[col].transform(lambda x: x.rolling(window=12, min_periods=1).mean())


df.head()
```

## Save Enhanced Dataset

```python

output_path = '../data/processed/final_3d_df_with_rolling_averages.csv'
df.to_csv(output_path, index=False)
```
